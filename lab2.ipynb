## Build a system for detecting anomalous patterns in text, which can be applied to identify unusual events or fraudulent activities in various domains.

```python
suspicious_keywords = [
    "unauthorized",
    "unknown location",
    "unexpected",
    "suspicious",
    "fraud",
    "stolen",
    "multiple failed logins",
    "unusual transaction"
]
```

```python
texts = [
    "Your monthly electricity bill is 850 rupees.",
    "Unauthorized transaction of 90000 rupees detected.",
    "Login detected from an unknown location.",
    "Your order will be delivered tomorrow.",
    "Multiple failed logins detected on your account.",
    "The meeting is scheduled for 10 AM.",
    "Suspicious activity detected in the bank account."
]
```

```python
print("Text Anomaly Detection System")
for text in texts:
    lower_text = text.lower()

    matched_keywords = [
        word for word in suspicious_keywords
        if word in lower_text
    ]

    if matched_keywords:
        status = "ANOMALY DETECTED"
        reason = ", ".join(matched_keywords)
    else:
        status = "NORMAL"
        reason = "No predefined suspicious keywords found"

    print("\nText:", text)
    print("Status:", status)
    print("Reason:", reason)
```
