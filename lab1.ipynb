## Develop a model that can effectively detect and understand sarcasm and irony in text, addressing the challenges posed by nuanced language.

```python
from transformers import pipeline
```

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
```

```python
labels = [
    "sarcastic or ironic",
    "literal or sincere"
]
```

```python
sentences = [
    "Oh great, another meeting that could have been an email.",
    "I just love waiting two hours for a delayed train.",
    "What a brilliant idea to go outside without an umbrella in a storm.",
    "Thank you for helping me complete the assignment.",
    "The weather is pleasant and sunny today.",
    "My phone battery died just before my online exam. Perfect timing!",
    "The fire station caught fire yesterday.",
    "She was genuinely happy with her examination results."
]
```

```python
print("Sarcasm and Irony Detection System")
for sentence in sentences:
    result = classifier(
        sentence,
        candidate_labels=labels,
        multi_label=False
    )

    predicted_label = result["labels"][0]
    confidence = result["scores"][0]

    print("\nSentence:", sentence)
    print("Prediction:", predicted_label)
    print("Confidence:", round(confidence * 100, 2), "%")
```
