## Create a sentiment analysis model that can effectively analyze sentiment in multilingual text, considering the nuances of different languages.

```python
from transformers import pipeline

model = pipeline(
    "text-classification",
    model="nlptown/bert-base-multilingual-uncased-sentiment",
    device=-1
)
```

```python
def analyze_sentiment(text):
    result = model(text)[0]
    stars = int(result["label"][0])

    if stars <= 2:
        sentiment = "Negative"
    elif stars == 3:
        sentiment = "Neutral"
    else:
        sentiment = "Positive"

    print("Text:", text)
    print("Sentiment:", sentiment)
    print("Rating:", result["label"])
    print("Confidence:", round(result["score"], 3))
```

```python
analyze_sentiment("The movie was excellent!")
analyze_sentiment("यह फिल्म बहुत अच्छी है।")
analyze_sentiment("ಈ ಸಿನಿಮಾ ತುಂಬಾ ಚೆನ್ನಾಗಿದೆ.")
analyze_sentiment("The movie was not that good!")
```
