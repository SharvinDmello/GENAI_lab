## Explore methods for zero-shot learning in text classification, where a model can classify samples from classes it has never seen during training.

```python
from transformers import pipeline
```

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
```

```python
sentence = "The new Tesla car has amazing battery life and fast charging."
```

```python
labels = [
    "technology",
    "sports",
    "politics",
    "entertainment"
]
```

```python
result = classifier(
    sentence,
    candidate_labels=labels
)
```

```python
print("Input:", sentence)
print("Classification:", result)
print("Predicted Category:", result["labels"][0])
print("Confidence Score:", round(result["scores"][0] * 100, 2), "%")
```
